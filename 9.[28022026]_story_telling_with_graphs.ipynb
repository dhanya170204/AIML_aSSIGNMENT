```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample dataset (you can replace with your own dataset)
data = {
    "Category": ["Food", "Transport", "Shopping", "Entertainment", "Bills"],
    "Amount": [5000, 2000, 3500, 1500, 3000]
}

df = pd.DataFrame(data)

# 1. Bar Chart
plt.figure()
plt.bar(df["Category"], df["Amount"])
plt.title("Monthly Expenses - Bar Chart")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.show()

# 2. Pie Chart
plt.figure()
plt.pie(df["Amount"], labels=df["Category"], autopct='%1.1f%%')
plt.title("Expense Distribution - Pie Chart")
plt.show()

# 3. Histogram
plt.figure()
plt.hist(df["Amount"])
plt.title("Expense Distribution - Histogram")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
#The data represents monthly expenses across different categories. From the bar chart, it is clear that Food has the highest spending, followed by Shopping and Bills. The pie chart shows that Food contributes the largest share of total expenses, indicating it is the major expense area. The histogram reveals how the expense values are distributed, showing most expenses fall in the mid-range.

#Overall, the data suggests that spending is highest on essential needs like food and bills, while entertainment has the lowest expenditure.
```
