```python
import re
import hashlib

# Function to hash password
def hash_password(password):
    return hashlib.sha256(password.encode()).hexdigest()

# Function to validate password strength
def is_strong_password(password):
    if len(password) < 8:
        return False, "Password must be at least 8 characters long."
    
    if not re.search(r"[A-Z]", password):
        return False, "Password must contain at least one uppercase letter."
    
    if not re.search(r"[a-z]", password):
        return False, "Password must contain at least one lowercase letter."
    
    if not re.search(r"[0-9]", password):
        return False, "Password must contain at least one digit."
    
    if not re.search(r"[!@#$%^&*(),.?\":{}|<>]", password):
        return False, "Password must contain at least one special character."
    
    return True, "Strong password."

# Registration
def register():
    while True:
        password = input("Create a password: ")
        valid, message = is_strong_password(password)
        print(message)
        
        if valid:
            confirm = input("Confirm password: ")
            if password == confirm:
                print("Registration successful!")
                return hash_password(password)
            else:
                print("Passwords do not match. Try again.")

# Login system
def login(stored_password_hash):
    attempts = 3
    while attempts > 0:
        password = input("Enter your password: ")
        if hash_password(password) == stored_password_hash:
            print("Login successful!")
            return True
        else:
            attempts -= 1
            print(f"Incorrect password. Attempts left: {attempts}")
    
    print("Too many failed attempts. Access denied.")
    return False

# Main program
def main():
    print("----- USER REGISTRATION -----")
    stored_password = register()
    
    print("\n----- USER LOGIN -----")
    login(stored_password)

# Run program
main()
```
