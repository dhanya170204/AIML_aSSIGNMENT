```python
# Storing data for 5 students using dictionary
students = {
    "Student1": {"name": "Asha", "marks": 85},
    "Student2": {"name": "Rahul", "marks": 92},
    "Student3": {"name": "Neha", "marks": 78},
    "Student4": {"name": "Arjun", "marks": 88},
    "Student5": {"name": "Priya", "marks": 95}
}

def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "F"

topper = max(students.values(), key=lambda x: x["marks"])

total_marks = sum(student["marks"] for student in students.values())
average = total_marks / len(students)

print("----- Student Details -----")
for student in students.values():
    grade = get_grade(student["marks"])
    print(f"Name: {student['name']}, Marks: {student['marks']}, Grade: {grade}")

print("\n----- Topper -----")
print(f"Topper: {topper['name']} with {topper['marks']} marks")

print("\n----- Class Average -----")
print(f"Average Marks: {average:.2f}")
```
