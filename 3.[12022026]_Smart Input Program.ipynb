```python
# Taking user input
name = input("Enter your name: ")
age = int(input("Enter your age: "))
hobby = input("Enter your favorite hobby: ")

# Categorizing age
if age < 13:
    category = "Child"
elif age < 20:
    category = "Teenager"
elif age < 60:
    category = "Adult"
else:
    category = "Senior"

# Personalized message
print("\n----- Personalized Message -----")
print(f"Hello {name}!")
print(f"You are a {category}.")
print(f"Your hobby is {hobby}, which is awesome! Keep enjoying it 😊")
```
