```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

# Download required dataset
nltk.download('vader_lexicon')
```

```python
sia = SentimentIntensityAnalyzer()
```

```python
reviews = [
    "The movie was absolutely fantastic with brilliant acting and storyline",
    "I hated the movie, it was boring and too long",
    "The film was okay, not great but not bad either",
    "Amazing visuals and great music, I really enjoyed it",
    "Worst movie ever, waste of time and money"
]
```

```python
def analyze_sentiment(review):
    score = sia.polarity_scores(review)
    
    if score['compound'] >= 0.05:
        return "Positive 😊"
    elif score['compound'] <= -0.05:
        return "Negative 😠"
    else:
        return "Neutral 😐"
```

```python
results = []

for review in reviews:
    sentiment = analyze_sentiment(review)
    scores = sia.polarity_scores(review)
    
    results.append({
        "Review": review,
        "Positive Score": scores['pos'],
        "Negative Score": scores['neg'],
        "Neutral Score": scores['neu'],
        "Compound Score": scores['compound'],
        "Final Sentiment": sentiment
    })
```

```python
import pandas as pd

df = pd.DataFrame(results)

df
```

```python
for i, row in df.iterrows():
    print(f"\nReview {i+1}: {row['Review']}")
    print(f"Sentiment: {row['Final Sentiment']}")
```
