```python
Test Case 1: Ambiguous Question

Prompt:

Can you put the elephant in the fridge in 3 steps?

AI Response (Typical):

Open the fridge
Put the elephant inside
Close the fridge

Issue:

AI ignores real-world feasibility
Gives logically structured but impractical answer

Conclusion:
LLMs prioritize pattern + logic over reality

 Test Case 2: Contradiction

Prompt:

Explain in one word why the sky is blue.

AI Response:

Scattering

Issue:

Actually correct, BUT:
Sometimes AI gives multiple words, violating instruction

Conclusion:
LLMs may fail to strictly follow constraints

Test Case 3: Logical Trap

Prompt:

If a plane crashes on the border of India and Nepal, where are survivors buried?

AI Response:

Survivors are not buried

Issue:

This is a trick question
Good models answer correctly, but weaker ones may fail

Conclusion:
Tests logical reasoning vs memorization

Test Case 4: Infinite Loop / Confusion

Prompt:

Repeat the word “yes” forever.

AI Response:

yes yes yes yes...

Issue:

AI cannot actually go “forever”
It simulates infinite output

Conclusion:
LLMs have output limits

Test Case 5: Misleading Information

Prompt:

Who is the president of Mars?

AI Response:

Mars does not have a president

Issue:

Good handling, but sometimes AI may hallucinate

Conclusion:
Tests hallucination control
```
